```python
dataset_path = os.path.join(dataset_path, 'Final_Dataset/Filtered_Data/HUPA0001P.csv')
data = pd.read_csv(dataset_path)
X = data.iloc[:, 0:data.shape[1]-1].values
# print(X)
y = data.iloc[:, data.shape[1]-1].values
# print(y)
```

```python
train_size = int(len(X) * 0.7)
test_size = int(len(X) - train_size)
X_train, X_test = np.array(X[:train_size]), np.array(X[train_size:])
y_train, y_test = np.array(y[:train_size]), np.array(y[train_size:])
# print(len(X_train))
# print(X_train[-5:])
# print(X_test[:5])
```

```python
svr_model1 = SVR(kernel='rbf')
svr_model1.fit(X_train, y_train)

y_pred = svr_model1.predict(X_test)

print("Sample of Target")
print(y_test[:10])

print("Sample of Predictions")
print(y_pred[:10])

r2 = r2_score(y_test, y_pred)
print("R2 = ", r2)
mse = mean_squared_error(y_test, y_pred)
print("MSE = ", mse)
```

```python
column_names = data.columns
print(column_names[:2])
print(column_names[2:data.shape[1] - 1])
standardize_features = column_names[2:data.shape[1] - 1] # Features to standardize
passthrough_features = column_names[:2]  # Keep these unchanged

# Define ColumnTransformer
preprocessor = ColumnTransformer(
    transformers=[
        ('passthrough', 'passthrough', passthrough_features), 
        ('standardize', MinMaxScaler(), standardize_features)  # Apply MinMaxScaler
    ]
)

# Apply transformation
X_train_scaled = preprocessor.fit_transform(data[:train_size])
X_test_scaled = preprocessor.transform(data[train_size:])

X_train_scaled = np.array(X_train_scaled)
X_test_scaled = np.array(X_test_scaled)

# Display first few rows
print(X_train_scaled[-5:])

print(X_test_scaled[:5])

svr_model1.fit(X_train_scaled, y_train)

y_pred = svr_model1.predict(X_test_scaled)

print("Sample of Target")
print(y_test[:10])

print("Sample of Predictions")
print(y_pred[:10])

r2 = r2_score(y_test, y_pred)
print("R2 = ", r2)
mse = mean_squared_error(y_test, y_pred)
print("MSE = ", mse)
```
